# ColumnTransformer in Scikit-learn

## What is ColumnTransformer?

ColumnTransformer applies different preprocessing techniques to different columns of a dataset.

Examples:
- Numerical columns: handle missing values and apply StandardScaler.
- Categorical columns: handle missing values and apply OneHotEncoder.

## Why do we use it?

It allows us to preprocess different column types together in a single transformer.


```python

import pandas as pd
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Sample dataset
df = pd.DataFrame({
    "Age": [22, 25, np.nan, 30, 28],
    "Fare": [100, 200, 150, np.nan, 300],
    "Sex": ["male", "female", "female", "male", np.nan],
    "Embarked": ["S", "C", "S", np.nan, "Q"]
})

X = df.copy()

# Numerical preprocessing
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),  #fills missing vals with median
    ("scaler", StandardScaler())
])

# Categorical preprocessing
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),  #fills missing vals
    ("encoder", OneHotEncoder(handle_unknown="ignore"))    #Converts categories into numerical features
])

# Combine both pipelines
ct = ColumnTransformer([
    ("num", num_pipeline, ["Age", "Fare"]),      #(name of new column, transformer type , column on which u apply this transformer)
    ("cat", cat_pipeline, ["Sex", "Embarked"])
])

X_transformed = ct.fit_transform(X)

print("Original data:")
print(X)

print("\nTransformed data:")
print(X_transformed)

print("\nOutput feature names:")
print(ct.get_feature_names_out())
```

```python

# Convert transformed data into tabular form(for reading purpose)

feature_names = ct.get_feature_names_out()

df_transformed = pd.DataFrame(
    X_transformed,
    columns=feature_names,
    index=df.index
)

df_transformed
```
